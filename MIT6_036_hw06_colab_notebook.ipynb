```
import numpy as np

def SM(z):
  # implement softmax
  return np.exp(z)/np.sum(np.exp(z))
```

 **Problem 2A**

```
z = np.array([[-1, 0, 1]]).T
a = SM(z)
print(a)
```

**Problem 2.C-F**

```
w = np.array([[1, -1, -2], [-1, 2, 1]])
x = np.array([[1], [1]])
y = np.array([[0, 1, 0]]).T

z = np.dot(w.T,x)
a = SM(z)

def NLL_grad_w(x,y,a):
    return x*(a-y).T



print(f'Q2C: {NLL_grad_w(x, y, a)}')


print(f'Q2D: {a[1,0]}')

## Q2E
step_size = 0.5
w_new = w - step_size*NLL_grad_w(x,y,a)
print(f'Q2E: {w_new}')

## Q2F
z_new = np.dot(w_new.T,x)
a_new = np.exp(z_new)/np.sum(np.exp(z_new))
print(f'Q2F: {a_new[1,0]}')
```
